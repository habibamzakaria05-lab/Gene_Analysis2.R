```python

# DNA Sequence Analysis

# 1. String
dna_sequence = "ATGCGTATCGGATCGATTACG"

# 2. Integer
sequence_length = len(dna_sequence)

# 3. Dictionary
nucleotide_counts = {
    "A": 0,
    "T": 0,
    "G": 0,
    "C": 0
}

# 4. List
valid_bases = ["A", "T", "G", "C"]

# 5. Count nucleotides using a for loop
for nucleotide in dna_sequence:
    if nucleotide in valid_bases:
        nucleotide_counts[nucleotide] += 1
    else:
        print("Invalid base found:", nucleotide)

# 6. Print counts using a while loop
keys = list(nucleotide_counts.keys())
index = 0

print("Nucleotide counts:")

while index < len(keys):
    nucleotide = keys[index]
    print(nucleotide, ":", nucleotide_counts[nucleotide])
    index += 1

# 7. Function to calculate GC content
def calculate_gc_content(sequence):
    g_count = sequence.count("G")
    c_count = sequence.count("C")

    gc_content = ((g_count + c_count) / len(sequence)) * 100
    return gc_content

# 8. Print results
gc_content = calculate_gc_content(dna_sequence)

print("Sequence length:", sequence_length)
print("GC content:", round(gc_content, 2), "%")
```
